```python
from datasets import load_dataset
from huggingface_hub import list_repo_files, hf_hub_download

import numpy as np
import pandas as pd
import json
import os
import re

REPO_ID = "witgaw/METR-LA"
REPO_TYPE = "dataset"
```

```python
repo_files = list_repo_files(
    repo_id=REPO_ID,
    repo_type=REPO_TYPE
)

print(f"Total files found: {len(repo_files)}\n")

for f in repo_files:
    print(f)

assert len(repo_files) > 0, "Repository contains no visible files."

# Discover required graph files by basename.
required_graph_files = [
    "adj_mx.npy",
    "adj_mx_mapping.json",
    "distances.csv",
    "sensor_locations.csv"
]

graph_paths = {}

for basename in required_graph_files:
    matches = [
        f for f in repo_files
        if os.path.basename(f) == basename
    ]

    print(f"\n{basename}: {matches}")

    assert len(matches) > 0, (
        f"{basename} not found anywhere in repository."
    )

    assert len(matches) == 1, (
        f"Expected exactly one {basename}, found: {matches}"
    )

    graph_paths[basename] = matches[0]

print("\nResolved graph paths:")
for name, path in graph_paths.items():
    print(f"{name:25s} -> {path}")
```

```python
train = load_dataset(
    REPO_ID,
    split="train"
)

print(train)

n_rows = train.num_rows
n_cols = len(train.column_names)

print("\nShape:", (n_rows, n_cols))

assert n_rows > 0, "Train split is empty."
assert n_cols > 0, "Train split has no columns."
```

```python
print("Features / dtypes:\n")

for col, feature in train.features.items():
    print(f"{col:25s} -> {feature}")

columns = train.column_names

print("\nFull column list:")
for i, col in enumerate(columns):
    print(f"{i:3d}: {col}")

assert "node_id" in columns, "node_id column missing."
assert "t0_timestamp" in columns, "t0_timestamp column missing."
```

```python
# ============================================================
# Split columns into ID / X inputs / Y targets
# ============================================================

import re

columns = train.column_names

id_cols = ["node_id", "t0_timestamp"]

x_cols = [c for c in columns if c.startswith("x_t")]
y_cols = [c for c in columns if c.startswith("y_t")]


def x_sort_key(col):
    """
    Handles:
        x_t-11_d0
        x_t-1_d1
        x_t+0_d0
        x_t+0_d1
    """
    m = re.fullmatch(r"x_t([+-]\d+)_d(\d+)", col)

    assert m is not None, f"Unexpected x column format: {col}"

    offset = int(m.group(1))
    dim = int(m.group(2))

    return offset, dim


def y_sort_key(col):
    """
    Handles:
        y_t+1_d0
        y_t+1_d1
        ...
    """
    m = re.fullmatch(r"y_t([+-]\d+)_d(\d+)", col)

    assert m is not None, f"Unexpected y column format: {col}"

    offset = int(m.group(1))
    dim = int(m.group(2))

    return offset, dim


x_cols = sorted(x_cols, key=x_sort_key)
y_cols = sorted(y_cols, key=y_sort_key)


# ------------------------------------------------------------
# Print groups
# ------------------------------------------------------------

print("ID COLUMNS:")
print(id_cols)

print("\nX INPUT COLUMNS:")
for c in x_cols:
    print(c)

print("\nY TARGET COLUMNS:")
for c in y_cols:
    print(c)


# ------------------------------------------------------------
# Assertions
# ------------------------------------------------------------

assert "node_id" in columns
assert "t0_timestamp" in columns

assert len(x_cols) > 0, "No x_t input columns detected."
assert len(y_cols) > 0, "No y_t target columns detected."

# METR-LA: 12 lookback positions × 2 dimensions
assert len(x_cols) == 24, (
    f"Expected 24 X columns, found {len(x_cols)}"
)

# Expected offsets:
# -11, -10, ..., -1, 0
expected_x = {
    f"x_t{offset:+d}_d{d}"
    for offset in range(-11, 1)
    for d in (0, 1)
}

assert set(x_cols) == expected_x, (
    "\nX column structure mismatch."
    f"\nMissing: {expected_x - set(x_cols)}"
    f"\nUnexpected: {set(x_cols) - expected_x}"
)

assert set(id_cols).isdisjoint(x_cols)
assert set(id_cols).isdisjoint(y_cols)
assert set(x_cols).isdisjoint(y_cols)

classified = set(id_cols) | set(x_cols) | set(y_cols)

assert classified == set(columns), (
    f"Unclassified columns: {set(columns) - classified}"
)

print("\n✓ Column structure validation PASSED")
```

```python
sample = train[0]

print("node_id     :", sample["node_id"])
print("t0_timestamp:", sample["t0_timestamp"])

print("\nInput values:")
for col in x_cols:
    print(f"{col:15s} = {sample[col]}")

assert sample["node_id"] is not None
assert sample["t0_timestamp"] is not None

print("\n✓ Sample row validation PASSED")
```

```python
t0 = pd.Timestamp(sample["t0_timestamp"])

assert t0.minute % 5 == 0, \
    f"Timestamp {t0} is not on a 5-minute boundary."

lags_to_test = [0, 1, 5, 11]

for lag in lags_to_test:

    offset = -lag
    col = f"x_t{offset:+d}_d1"

    assert col in columns, f"Missing column: {col}"

    historical_time = t0 - pd.Timedelta(minutes=5 * lag)

    step = (
        historical_time.hour * 12
        + historical_time.minute // 5
    )

    expected = step / 288.0
    actual = float(sample[col])

    print(f"\n{col}")
    print("Time     :", historical_time)
    print("Step     :", step)
    print("Expected :", expected)
    print("Actual   :", actual)

    assert np.isclose(actual, expected, atol=1e-6), (
        f"{col}: expected {expected}, got {actual}"
    )

print("\n✓ d1 TIME-OF-DAY TEST PASSED")
```

```python
rows_to_check = [
    0,
    len(train) // 4,
    len(train) // 2,
    3 * len(train) // 4,
    len(train) - 1
]

for idx in rows_to_check:

    row = train[idx]
    t0 = pd.Timestamp(row["t0_timestamp"])

    assert t0.minute % 5 == 0

    for lag in [0, 1, 5, 11]:

        col = f"x_t{-lag:+d}_d1"

        historical_time = (
            t0 - pd.Timedelta(minutes=5 * lag)
        )

        step = (
            historical_time.hour * 12
            + historical_time.minute // 5
        )

        expected = step / 288.0
        actual = float(row[col])

        assert np.isclose(
            actual,
            expected,
            atol=1e-6
        ), (
            f"Row {idx}, {col}: "
            f"expected {expected}, got {actual}"
        )

print("✓ Multi-row d1 validation PASSED")
```

```python
node_ids = np.asarray(train["node_id"])
timestamps = np.asarray(train["t0_timestamp"])

unique_nodes = np.unique(node_ids)
unique_timestamps = np.unique(timestamps)

n_sensors = len(unique_nodes)
n_timestamps = len(unique_timestamps)

actual_rows = len(train)
expected_rows = n_sensors * n_timestamps

print("Sensors          :", n_sensors)
print("Timestamps       :", n_timestamps)
print("Expected rows    :", expected_rows)
print("Actual rows      :", actual_rows)

assert n_sensors in (207, 325), (
    f"Unexpected sensor count: {n_sensors}"
)

assert actual_rows == expected_rows, (
    f"Row mismatch: {actual_rows} != {expected_rows}"
)

print("\n✓ SENSOR × TIMESTAMP TEST PASSED")
```

```python
pairs = set(zip(node_ids.tolist(), timestamps.tolist()))

duplicate_count = actual_rows - len(pairs)

print("Duplicate (node_id, timestamp) pairs:",
      duplicate_count)

assert duplicate_count == 0, (
    f"Found {duplicate_count} duplicate pairs"
)

print("✓ DUPLICATE TEST PASSED")
```

```python
local_files = {}

for basename, repo_path in graph_paths.items():

    print("Downloading:", repo_path)

    local_path = hf_hub_download(
        repo_id=REPO_ID,
        filename=repo_path,
        repo_type="dataset"
    )

    assert os.path.isfile(local_path), (
        f"Failed to download {basename}"
    )

    local_files[basename] = local_path

print("\n✓ ALL GRAPH FILES DOWNLOADED")
```

```python
adj_mx = np.load(
    local_files["adj_mx.npy"],
    allow_pickle=False
)

print("Shape :", adj_mx.shape)
print("dtype :", adj_mx.dtype)

assert isinstance(adj_mx, np.ndarray)
assert adj_mx.ndim == 2
assert adj_mx.shape[0] == adj_mx.shape[1]

assert adj_mx.shape[0] in (207, 325), (
    f"Unexpected adjacency size: {adj_mx.shape}"
)

assert adj_mx.shape[0] == n_sensors, (
    f"Adjacency has {adj_mx.shape[0]} nodes "
    f"but dataset has {n_sensors}"
)

print("\n✓ ADJACENCY MATRIX TEST PASSED")
```

```python
with open(
    local_files["adj_mx_mapping.json"],
    "r",
    encoding="utf-8"
) as f:
    mapping = json.load(f)

print("Mapping type:", type(mapping))

print("\nRaw structure:")
print(json.dumps(mapping, indent=2))

assert isinstance(mapping, (dict, list)), (
    f"Unexpected mapping type: {type(mapping)}"
)

assert len(mapping) > 0, "Mapping is empty"

print("\n✓ MAPPING STRUCTURE TEST PASSED")
```

```python
distances = pd.read_csv(
    local_files["distances.csv"]
)

locations = pd.read_csv(
    local_files["sensor_locations.csv"]
)

print("distances.csv")
print("Shape:", distances.shape)
print("Columns:", distances.columns.tolist())
display(distances.head())

print("\nsensor_locations.csv")
print("Shape:", locations.shape)
print("Columns:", locations.columns.tolist())
display(locations.head())

assert not distances.empty, "distances.csv is empty"
assert not locations.empty, "sensor_locations.csv is empty"

assert len(distances.columns) > 0
assert len(locations.columns) > 0

print("\n✓ GRAPH METADATA TEST PASSED")
```

```python
node_ids_int = np.asarray(unique_nodes, dtype=int)

N = adj_mx.shape[0]

print("node_id range :", node_ids_int.min(),
      "to", node_ids_int.max())

print("adj_mx range  : 0 to", N - 1)

assert node_ids_int.min() >= 0

assert node_ids_int.max() < N, (
    "node_id exceeds adjacency matrix range"
)

assert len(node_ids_int) == N, (
    "Number of node IDs does not match adjacency size"
)

expected_indices = np.arange(N)

assert np.array_equal(
    np.sort(node_ids_int),
    expected_indices
), (
    "node_id values are not exactly 0..N-1"
)

print("\n✓ NODE_ID ↔ ADJACENCY ALIGNMENT PASSED")
```

```python
# Cell 17 — Validate adjacency metadata mapping

print("Mapping type :", type(mapping).__name__)
print("Mapping keys :", list(mapping.keys()))
print("Graph size   :", N)

# Validate structure
assert isinstance(mapping, dict), (
    f"Expected mapping metadata to be dict, got {type(mapping)}"
)

assert "sensor_ids" in mapping, "sensor_ids missing from mapping"
assert "shape" in mapping, "shape missing from mapping"

sensor_ids = mapping["sensor_ids"]

print("Sensor IDs   :", len(sensor_ids))
print("Stored shape :", mapping["shape"])

print("\nFirst 10 sensor IDs:")
for i, sensor_id in enumerate(sensor_ids[:10]):
    print(i, "->", sensor_id)


# ------------------------------------------------------------
# Assertions
# ------------------------------------------------------------

assert isinstance(sensor_ids, list), (
    "mapping['sensor_ids'] must be a list"
)

assert len(sensor_ids) == N, (
    f"Mapping contains {len(sensor_ids)} sensor IDs, "
    f"but graph contains {N} nodes"
)

assert len(set(sensor_ids)) == N, (
    "Duplicate physical sensor IDs found"
)

assert mapping["shape"] == [N, N], (
    f"Stored graph shape {mapping['shape']} "
    f"does not match actual {adj_mx.shape}"
)

assert len(sensor_ids) == n_sensors, (
    f"Mapping has {len(sensor_ids)} sensors "
    f"but train has {n_sensors}"
)

print("\n✓ SENSOR MAPPING TEST PASSED")
```

```python
# Cell 18 — Final validation summary

tests = {
    "Correct sensor count":
        n_sensors in (207, 325),

    "Complete sensor x timestamp grid":
        actual_rows == n_sensors * n_timestamps,

    "Adjacency matrix is square":
        adj_mx.ndim == 2
        and adj_mx.shape[0] == adj_mx.shape[1],

    "Sensor count matches adjacency":
        n_sensors == adj_mx.shape[0],

    "node_id matches adjacency indices":
        np.array_equal(
            np.sort(node_ids_int),
            np.arange(adj_mx.shape[0])
        ),

    "Sensor mapping matches adjacency":
        len(mapping["sensor_ids"]) == adj_mx.shape[0],

    "Stored adjacency shape matches actual":
        mapping["shape"] == list(adj_mx.shape),

    "Physical sensor IDs are unique":
        len(set(mapping["sensor_ids"]))
        == len(mapping["sensor_ids"]),

    "Distances file loaded":
        not distances.empty,

    "Sensor locations file loaded":
        not locations.empty
}


print("=" * 60)
print("FINAL VALIDATION RESULTS")
print("=" * 60)

for test, passed in tests.items():
    status = "PASS" if passed else "FAIL"
    print(f"{status:4} : {test}")

print("=" * 60)


if all(tests.values()):

    print("\nALL TEST CASES PASSED")

    print(f"""
Dataset             : {REPO_ID}
Train rows          : {actual_rows:,}
Unique sensors      : {n_sensors}
Unique timestamps   : {n_timestamps:,}
Input columns       : {len(x_cols)}
Target columns      : {len(y_cols)}
Adjacency shape     : {adj_mx.shape}
Physical sensor IDs : {len(mapping["sensor_ids"])}

METR-LA dataset and sensor graph are correctly aligned.
""")

else:

    failed_tests = [
        test
        for test, passed in tests.items()
        if not passed
    ]

    print("\nVALIDATION FAILED")

    print("\nFailed tests:")

    for test in failed_tests:
        print(" -", test)

    raise AssertionError(
        f"{len(failed_tests)} validation test(s) failed."
    )
```
